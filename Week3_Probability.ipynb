## 1. Probability Fundamentals

Probability measures how likely an event is to occur.

The probability of an event \(A\) is written as:

\[
P(A)
\]

Probability values range from 0 to 1.

- \(P(A)=0\): the event is impossible under the model.
- \(P(A)=1\): the event is certain under the model.
- \(0<P(A)<1\): the event has a probability between impossible and certain.

Probability can also be expressed as a percentage. For example, \(0.75=75\%\).

### 1.1 Random experiments

A random experiment is a process whose outcome is uncertain before it takes place.

Examples include:

- Tossing a coin.
- Rolling a die.
- Selecting a card from a deck.
- Randomly selecting a customer from a dataset.
- Observing whether a transaction is fraudulent.

The outcome is not necessarily unpredictable in every sense, but it is treated as uncertain within the probability model.

### 1.2 Sample space

The sample space is the set of all possible outcomes of a random experiment. It is usually represented by \(S\) or \(\Omega\).

For a coin toss:

\[
S=\{H,T\}
\]

Where \(H\) represents heads and \(T\) represents tails.

For a six-sided die:

\[
S=\{1,2,3,4,5,6\}
\]

### 1.3 Events

An event is a set of outcomes from the sample space.

For example, when rolling a die, let \(A\) be the event of rolling an even number.

\[
A=\{2,4,6\}
\]

Assuming a fair die, the probability of this event is:

\[
P(A)=\frac{3}{6}=\frac{1}{2}
\]

### 1.4 Equally likely outcomes

When all outcomes are equally likely, the probability of an event can be calculated as:

\[
\boxed{
P(A)=\frac{\text{Number of favourable outcomes}}
{\text{Total number of possible outcomes}}
}
\]

For example, the probability of rolling a number greater than 4 on a fair six-sided die is:

\[
P(X>4)=\frac{2}{6}=\frac{1}{3}
\]

The favourable outcomes are 5 and 6.

**Important:** This formula applies when the possible outcomes are equally likely. It should not automatically be used for biased coins, unequal sampling probabilities, or other situations where outcomes have different probabilities.

## 2. Basic Probability Rules

### 2.1 The complement rule

The complement of an event \(A\), written as \(A^c\), is the event that \(A\) does not occur.

The complement rule is:

\[
\boxed{P(A^c)=1-P(A)}
\]

For example, suppose the probability of a model classifying a transaction as fraudulent is 0.08.

The probability that it is not classified as fraudulent is:

\[
P(A^c)=1-0.08=0.92
\]

Therefore, the complementary probability is 0.92, or 92%.

This rule is useful when calculating the probability of something not happening.

### 2.2 The addition rule

The addition rule helps calculate the probability that event \(A\), event \(B\), or both occur.

The general formula is:

\[
\boxed{
P(A\cup B)=P(A)+P(B)-P(A\cap B)
}
\]

Where:

- \(A\cup B\) means \(A\) or \(B\) or both occur.
- \(A\cap B\) means both \(A\) and \(B\) occur.

We subtract the intersection because it would otherwise be counted twice.

#### Example

Suppose that when rolling a fair die:

- \(A\) is the event of rolling an even number: \(\{2,4,6\}\).
- \(B\) is the event of rolling a number greater than 3: \(\{4,5,6\}\).

Then:

\[
P(A)=\frac{3}{6},\qquad P(B)=\frac{3}{6}
\]

Their intersection is:

\[
A\cap B=\{4,6\}
\]

Therefore:

\[
P(A\cap B)=\frac{2}{6}
\]

Applying the addition rule:

\[
P(A\cup B)=\frac{3}{6}+\frac{3}{6}-\frac{2}{6}
\]

\[
\boxed{P(A\cup B)=\frac{4}{6}=\frac{2}{3}}
\]

### 2.3 Mutually exclusive events

Two events are mutually exclusive if they cannot occur at the same time.

For example, on a single coin toss, getting heads and getting tails are mutually exclusive.

For mutually exclusive events:

\[
P(A\cap B)=0
\]

The addition rule becomes:

\[
\boxed{P(A\cup B)=P(A)+P(B)}
\]

This simplified rule should only be used when the events cannot overlap.

### 2.4 The multiplication rule

The general multiplication rule is:

\[
\boxed{P(A\cap B)=P(A)P(B\mid A)}
\]

This means that the probability of both events occurring is the probability of \(A\) multiplied by the probability of \(B\) given that \(A\) has occurred.

If \(A\) and \(B\) are independent, then \(P(B\mid A)=P(B)\), giving:

\[
\boxed{P(A\cap B)=P(A)P(B)}
\]

For example, when tossing a fair coin twice, the probability of getting heads on both tosses is:

\[
P(HH)=\frac{1}{2}\times\frac{1}{2}
\]

\[
\boxed{P(HH)=\frac{1}{4}}
\]

The multiplication rule for independent events is useful when modelling repeated trials and combinations of events.

## 3. Conditional Probability

Conditional probability measures the probability of an event occurring given that another event has already occurred.

It is written as:

\[
P(A\mid B)
\]

This is read as "the probability of \(A\) given \(B\)."

The formula is:

\[
\boxed{
P(A\mid B)=\frac{P(A\cap B)}{P(B)}
}
\]

provided that \(P(B)>0\).

### Example: Rolling a die

Suppose a fair six-sided die is rolled. We know that the result is greater than 3. What is the probability that the result is even?

Define:

\[
A=\{2,4,6\}
\]

\[
B=\{4,5,6\}
\]

Since we know the result is greater than 3, our possible outcomes are now only:

\[
B=\{4,5,6\}
\]

Among these three outcomes, 4 and 6 are even.

Therefore:

\[
P(A\mid B)=\frac{2}{3}
\]

We can also use the conditional probability formula:

\[
P(A\cap B)=\frac{2}{6}
\]

\[
P(B)=\frac{3}{6}
\]

Thus:

\[
P(A\mid B)=\frac{2/6}{3/6}
=\boxed{\frac{2}{3}}
\]

The important idea is that knowing \(B\) occurred changes the set of outcomes we consider.

### Conditional probability in Machine Learning

Suppose we want to estimate the probability that an email is spam given that it contains a suspicious link.

We write:

\[
P(\text{Spam}\mid\text{Suspicious Link})
\]

This probability may differ from the overall probability that an email is spam.

Conditional probabilities help models use observed features to estimate the likelihood of different outcomes.

## 4. Independent and Dependent Events

### 4.1 Independent events

Two events are independent if the occurrence of one does not change the probability of the other.

Mathematically:

\[
\boxed{P(A\mid B)=P(A)}
\]

when the conditional probability is defined.

An equivalent condition is:

\[
\boxed{P(A\cap B)=P(A)P(B)}
\]

For example, consider two tosses of a fair coin. The result of the first toss does not affect the result of the second toss, assuming the coin tosses are conducted independently.

The probability of heads on the second toss remains:

\[
P(H_2)=\frac{1}{2}
\]

even if the first toss resulted in heads.

### 4.2 Dependent events

Two events are dependent if knowing that one has occurred changes the probability of the other.

For example, suppose a bag contains three red balls and two blue balls. Two balls are selected without replacement.

The probability that the first ball is red is:

\[
P(R_1)=\frac{3}{5}
\]

If the first ball is red, only two red balls remain among four balls. Therefore:

\[
P(R_2\mid R_1)=\frac{2}{4}=\frac{1}{2}
\]

The probability of a red second ball changes after observing a red first ball. The events are dependent.

The probability of drawing two red balls is:

\[
P(R_1\cap R_2)=\frac{3}{5}\times\frac{2}{4}
\]

\[
\boxed{P(R_1\cap R_2)=\frac{3}{10}}
\]

### Independence versus mutual exclusivity

These concepts are different.

- **Independent events:** One event does not change the probability of the other.
- **Mutually exclusive events:** The two events cannot occur together.

Two mutually exclusive events with positive probabilities cannot be independent. If one occurs, the other becomes impossible.

## 5. Bayes' Theorem

Bayes' theorem allows us to update a probability when new information becomes available.

It is especially useful when we know the probability of observing some evidence under different conditions and want to calculate the probability of a condition given that evidence.

The formula is:

\[
\boxed{
P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}
}
\]

Where:

- \(P(A)\) is the prior probability of \(A\).
- \(P(B\mid A)\) is the likelihood of observing \(B\) if \(A\) is true.
- \(P(B)\) is the overall probability of observing \(B\).
- \(P(A\mid B)\) is the posterior probability of \(A\) after observing \(B\).

### 5.1 Understanding the terms

**Prior:** What we believe about the probability of an event before considering the new evidence.

**Likelihood:** How likely the observed evidence is if the event is true.

**Evidence:** The overall probability of observing the evidence under all relevant possibilities.

**Posterior:** The updated probability after considering the evidence.

### 5.2 Medical test example

Suppose a condition affects 1% of a population.

A test has the following characteristics:

- Sensitivity: 99%. If someone has the condition, the test is positive 99% of the time.
- Specificity: 95%. If someone does not have the condition, the test is negative 95% of the time.

We want to calculate the probability that a person has the condition given that their test is positive.

Let:

- \(D\) mean the person has the condition.
- \(+\) mean the test result is positive.

The known probabilities are:

\[
P(D)=0.01
\]

\[
P(\neg D)=0.99
\]

\[
P(+\mid D)=0.99
\]

Since specificity is 95%, the false-positive rate is:

\[
P(+\mid\neg D)=1-0.95=0.05
\]

First, calculate the overall probability of a positive test:

\[
P(+)=P(+\mid D)P(D)+P(+\mid\neg D)P(\neg D)
\]

Substitute the values:

\[
P(+)=(0.99)(0.01)+(0.05)(0.99)
\]

\[
P(+)=0.0099+0.0495=0.0594
\]

Now apply Bayes' theorem:

\[
P(D\mid +)=\frac{P(+\mid D)P(D)}{P(+)}
\]

\[
P(D\mid +)=\frac{0.99\times0.01}{0.0594}
\]

\[
\boxed{P(D\mid +)\approx0.1667}
\]

Therefore, the probability is approximately 16.67%.

Even with a positive result, the probability that the person has the condition is not 99%. This is because the condition is rare, so false positives from the much larger group without the condition have a substantial effect.

This example illustrates why prior probabilities matter when interpreting evidence.

### 5.3 Bayes' theorem in Machine Learning

Bayes' theorem is useful in probabilistic classification and Bayesian learning.

For example, a spam classifier may estimate:

\[
P(\text{Spam}\mid\text{Words in Email})
\]

It uses information about how frequently particular words occur in spam and non-spam emails to estimate the probability that a message belongs to a class.

Naive Bayes classifiers use Bayes' theorem together with a simplifying assumption that features are conditionally independent given the class.

## 6. Random Variables

A random variable assigns a numerical value to each outcome of a random experiment.

It is commonly represented by \(X\), \(Y\), or another capital letter.

For example, suppose we toss two fair coins and let \(X\) be the number of heads.

The possible outcomes are:

| Outcome | Value of \(X\) |
|---|---:|
| TT | 0 |
| HT | 1 |
| TH | 1 |
| HH | 2 |

Therefore:

\[
X\in\{0,1,2\}
\]

A random variable makes it easier to describe and analyse uncertain numerical quantities.

### 6.1 Discrete random variables

A discrete random variable has a finite or countably infinite set of possible values.

Examples include:

- The number of heads in three coin tosses.
- The number of defective products in a batch.
- The number of customers who visit a store in an hour.

A discrete random variable can be described using a probability mass function.

### 6.2 Continuous random variables

A continuous random variable can take values over a continuous range.

Examples include:

- A person's height.
- The time taken to complete a task.
- The measured temperature of a room.

Continuous random variables are often described using a probability density function.

For a continuous random variable, the probability of one exact value is zero under an ordinary continuous distribution:

\[
P(X=a)=0
\]

Instead, probabilities are calculated over intervals, such as \(P(3\leq X\leq 7)\).

### 6.3 Probability mass function (PMF)

A probability mass function gives the probability of each possible value of a discrete random variable.

It is written as:

\[
p_X(x)=P(X=x)
\]

A valid PMF satisfies:

\[
p_X(x)\geq0
\]

and:

\[
\sum_x p_X(x)=1
\]

For the number of heads in two fair coin tosses:

| \(x\) | \(P(X=x)\) |
|---:|---:|
| 0 | \(1/4\) |
| 1 | \(1/2\) |
| 2 | \(1/4\) |

The probabilities sum to one:

\[
\frac14+\frac12+\frac14=1
\]

### 6.4 Probability density function (PDF)

A probability density function describes the density of probability for a continuous random variable.

It is written as:

\[
f_X(x)
\]

A valid PDF is non-negative, and the total area under its curve equals one:

\[
\int_{-\infty}^{\infty}f_X(x)\,dx=1
\]

The probability that \(X\) lies between \(a\) and \(b\) is:

\[
\boxed{
P(a\leq X\leq b)=\int_a^b f_X(x)\,dx
}
\]

A PDF's value at a point is not itself the probability of that exact point.

### 6.5 Cumulative distribution function (CDF)

The cumulative distribution function gives the probability that a random variable is less than or equal to a value \(x\).

\[
\boxed{F_X(x)=P(X\leq x)}
\]

The CDF works for both discrete and continuous random variables.

For example, if \(X\) is the number of heads in two fair coin tosses:

\[
P(X\leq1)=P(X=0)+P(X=1)
\]

\[
P(X\leq1)=\frac14+\frac12=\frac34
\]

Therefore:

\[
F_X(1)=0.75
\]

### 6.6 Joint and marginal distributions

When working with more than one random variable, we may need to describe their probabilities together.

A **joint distribution** describes the probabilities of combinations of values for two or more random variables.

For discrete random variables \(X\) and \(Y\):

\[
p_{X,Y}(x,y)=P(X=x,Y=y)
\]

A **marginal distribution** describes the probabilities of one variable without specifying the value of the other.

For example:

\[
P(X=x)=\sum_y P(X=x,Y=y)
\]

In words, we add the joint probabilities over all possible values of \(Y\) to obtain the marginal probability of \(X\).

Joint and marginal distributions are useful when studying relationships between features and outcomes in probabilistic models.

## 7. Expected Value, Variance, and Standard Deviation

These concepts describe important properties of random variables.

- Expected value describes the long-run average.
- Variance describes the spread of values around the expected value.
- Standard deviation expresses spread in the original units of the variable.

### 7.1 Expected value

The expected value is the probability-weighted average of the possible values of a random variable.

For a discrete random variable:

\[
\boxed{E[X]=\sum_x xP(X=x)}
\]

For a continuous random variable with density \(f_X(x)\):

\[
E[X]=\int_{-\infty}^{\infty}xf_X(x)\,dx
\]

#### Example: Two coin tosses

Let \(X\) be the number of heads in two fair coin tosses.

The distribution is:

| \(X\) | Probability |
|---:|---:|
| 0 | \(1/4\) |
| 1 | \(1/2\) |
| 2 | \(1/4\) |

The expected value is:

\[
E[X]=0\left(\frac14\right)+1\left(\frac12\right)+2\left(\frac14\right)
\]

\[
E[X]=0+\frac12+\frac12
\]

\[
\boxed{E[X]=1}
\]

This does not mean that every pair of coin tosses produces exactly one head. It means that the average number of heads approaches one over many repetitions under the same conditions.

### 7.2 Variance

Variance measures how far values tend to spread around the expected value.

It is defined as:

\[
\boxed{\operatorname{Var}(X)=E[(X-\mu)^2]}
\]

where:

\[
\mu=E[X]
\]

An equivalent formula is:

\[
\boxed{\operatorname{Var}(X)=E[X^2]-(E[X])^2}
\]

For a discrete random variable:

\[
E[X^2]=\sum_x x^2P(X=x)
\]

#### Example: Two coin tosses

We already know:

\[
E[X]=1
\]

Calculate the expected squared value:

\[
E[X^2]=0^2\left(\frac14\right)+1^2\left(\frac12\right)+2^2\left(\frac14\right)
\]

\[
E[X^2]=0+\frac12+1=1.5
\]

Now calculate the variance:

\[
\operatorname{Var}(X)=1.5-1^2
\]

\[
\boxed{\operatorname{Var}(X)=0.5}
\]

### 7.3 Standard deviation

Standard deviation is the square root of variance:

\[
\boxed{\sigma=\sqrt{\operatorname{Var}(X)}}
\]

For the two-coin example:

\[
\sigma=\sqrt{0.5}
\]

\[
\boxed{\sigma\approx0.7071}
\]

Unlike variance, standard deviation is expressed in the same units as the original variable.

### 7.4 Why these measures matter in Machine Learning

Expected values help describe average outcomes, while variance and standard deviation describe uncertainty and variability.

These ideas appear in probability models, statistical learning, uncertainty estimation, and methods for analysing model predictions.

Variance is also important in understanding the bias-variance trade-off, although that topic involves additional concepts beyond probability alone.

## 8. Bernoulli Distribution

The Bernoulli distribution models a single random trial with exactly two possible outcomes, usually represented by 1 and 0.

For example:

- Success = 1.
- Failure = 0.

The probability of success is \(p\), while the probability of failure is \(1-p\).

The PMF is:

\[
\boxed{
P(X=x)=p^x(1-p)^{1-x},\quad x\in\{0,1\}
}
\]

Equivalently:

\[
P(X=1)=p
\]

\[
P(X=0)=1-p
\]

### 8.1 Expected value

For a Bernoulli random variable:

\[
E[X]=0(1-p)+1(p)
\]

Therefore:

\[
\boxed{E[X]=p}
\]

### 8.2 Variance

Since \(X\) can only be 0 or 1, \(X^2=X\). Therefore:

\[
E[X^2]=p
\]

Using the variance formula:

\[
\operatorname{Var}(X)=E[X^2]-(E[X])^2
\]

\[
\operatorname{Var}(X)=p-p^2
\]

Hence:

\[
\boxed{\operatorname{Var}(X)=p(1-p)}
\]

### 8.3 Example

Suppose the probability that a particular transaction is fraudulent is 0.3.

Let:

\[
X=
\begin{cases}
1,&\text{if the transaction is fraudulent}\\
0,&\text{otherwise}
\end{cases}
\]

Then:

\[
P(X=1)=0.3
\]

\[
P(X=0)=1-0.3=0.7
\]

The expected value is:

\[
E[X]=0.3
\]

The variance is:

\[
\operatorname{Var}(X)=0.3(0.7)=0.21
\]

Therefore:

\[
\boxed{E[X]=0.3,\qquad\operatorname{Var}(X)=0.21}
\]

The expected value is a probability-weighted average of the numerical outcomes, not a claim that the transaction is partially fraudulent.

### 8.4 Python example

```python
from scipy.stats import bernoulli

p = 0.3

# Probability of success and failure
print("P(X = 1):", bernoulli.pmf(1, p))
print("P(X = 0):", bernoulli.pmf(0, p))

# Expected value and variance
print("Expected value:", bernoulli.mean(p))
print("Variance:", bernoulli.var(p))
```

Expected results:

```text
P(X = 1): 0.3
P(X = 0): 0.7
Expected value: 0.3
Variance: 0.21
```

### 8.5 Bernoulli distribution in Machine Learning

Bernoulli distributions can model binary outcomes such as:

- Whether an email is spam.
- Whether a transaction is fraudulent.
- Whether a customer clicks an advertisement.
- Whether a binary classification outcome is positive or negative.

A model may predict a probability \(p\), and a Bernoulli random variable represents the corresponding binary outcome.

## 9. Binomial Distribution

The Binomial distribution models the number of successes in a fixed number of independent Bernoulli trials, where each trial has the same probability of success.

For example, it can model the number of heads in ten independent tosses of a fair coin.

The random variable is written as:

\[
X\sim\operatorname{Binomial}(n,p)
\]

Where:

- \(n\) is the number of trials.
- \(p\) is the probability of success on each trial.
- \(X\) is the number of successes.

The PMF is:

\[
\boxed{
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
}
\]

for \(k=0,1,\ldots,n\).

The combination term is:

\[
\binom{n}{k}=\frac{n!}{k!(n-k)!}
\]

It counts the number of ways to choose which \(k\) trials are successes.

### 9.1 Expected value and variance

For a Binomial random variable:

\[
\boxed{E[X]=np}
\]

\[
\boxed{\operatorname{Var}(X)=np(1-p)}
\]

The standard deviation is:

\[
\boxed{\sigma=\sqrt{np(1-p)}}
\]

### 9.2 Example: Number of successes

Suppose an experiment consists of five independent trials, each with a success probability of 0.5.

Let:

\[
X\sim\operatorname{Binomial}(5,0.5)
\]

Find the probability of exactly two successes.

Using the Binomial PMF:

\[
P(X=2)=\binom52(0.5)^2(0.5)^3
\]

Since:

\[
\binom52=\frac{5!}{2!3!}=10
\]

Then:

\[
P(X=2)=10(0.5)^5
\]

\[
\boxed{P(X=2)=0.3125}
\]

The expected number of successes is:

\[
E[X]=5(0.5)=2.5
\]

The variance is:

\[
\operatorname{Var}(X)=5(0.5)(0.5)=1.25
\]

### 9.3 Python example

```python
from scipy.stats import binom

n = 5
p = 0.5

# Probability of exactly two successes
print("P(X = 2):", binom.pmf(2, n, p))

# Probability of at most two successes
print("P(X <= 2):", binom.cdf(2, n, p))

# Expected value and variance
print("Expected value:", binom.mean(n, p))
print("Variance:", binom.var(n, p))
```

The Binomial distribution is useful for modelling counts of binary outcomes, such as the number of successful predictions in a fixed set of independent trials, provided the assumptions are reasonable.

## 10. Uniform Distribution

A Uniform distribution assigns equal probability density to all values within a specified interval.

For a continuous Uniform random variable:

\[
X\sim U(a,b)
\]

where \(a\) is the lower bound and \(b\) is the upper bound, with \(a<b\).

Its PDF is:

\[
\boxed{
f_X(x)=
\begin{cases}
\frac{1}{b-a},&a\leq x\leq b\\
0,&\text{otherwise}
\end{cases}
}
\]

The density is constant throughout the interval.

### 10.1 Probability over an interval

For \(a\leq c\leq d\leq b\):

\[
\boxed{P(c\leq X\leq d)=\frac{d-c}{b-a}}
\]

For example, if:

\[
X\sim U(0,10)
\]

then:

\[
P(3\leq X\leq7)=\frac{7-3}{10-0}
\]

\[
\boxed{P(3\leq X\leq7)=0.4}
\]

The interval from 3 to 7 covers 4 units out of the full 10-unit range.

### 10.2 Expected value and variance

For a continuous Uniform distribution:

\[
\boxed{E[X]=\frac{a+b}{2}}
\]

\[
\boxed{\operatorname{Var}(X)=\frac{(b-a)^2}{12}}
\]

For \(X\sim U(0,10)\):

\[
E[X]=\frac{0+10}{2}=5
\]

\[
\operatorname{Var}(X)=\frac{(10-0)^2}{12}
=\frac{100}{12}
\approx8.3333
\]

The standard deviation is approximately:

\[
\sigma=\sqrt{8.3333}\approx2.8868
\]

### 10.3 Python example

```python
from scipy.stats import uniform

a = 0
b = 10

# SciPy's uniform distribution uses loc=a and scale=b-a
distribution = uniform(loc=a, scale=b - a)

print("P(3 <= X <= 7):", distribution.cdf(7) - distribution.cdf(3))
print("Expected value:", distribution.mean())
print("Variance:", distribution.var())
```

The Uniform distribution is useful when all values within a particular interval are treated as equally likely. It is also useful in simulations and in generating random values for experiments.

## 11. Normal Distribution

The Normal distribution is a continuous probability distribution with a symmetric, bell-shaped curve.

It is often used to model quantities that tend to cluster around an average, although real-world data does not always follow a Normal distribution.

A Normal random variable is written as:

\[
X\sim N(\mu,\sigma^2)
\]

Where:

- \(\mu\) is the mean.
- \(\sigma^2\) is the variance.
- \(\sigma\) is the standard deviation.

The PDF is:

\[
\boxed{
f_X(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
}
\]

for \(\sigma>0\).

### 11.1 Characteristics

The Normal distribution has several important properties:

- It is symmetric about its mean.
- Its mean, median, and mode are equal.
- The total area under its curve is 1.
- Its mean determines the centre of the distribution.
- Its standard deviation determines how spread out the curve is.

For a Normal distribution, approximately:

- 68% of observations fall within one standard deviation of the mean.
- 95% fall within two standard deviations.
- 99.7% fall within three standard deviations.

These percentages are approximations and assume the distribution is Normal.

### 11.2 The standard Normal distribution

The standard Normal distribution has:

\[
\mu=0,\qquad\sigma=1
\]

It is commonly represented by:

\[
Z\sim N(0,1)
\]

A value from another Normal distribution can be standardized using:

\[
\boxed{Z=\frac{X-\mu}{\sigma}}
\]

This converts a value into the number of standard deviations it lies above or below the mean.

For example, suppose:

\[
X\sim N(100,15^2)
\]

For \(X=130\):

\[
Z=\frac{130-100}{15}=2
\]

The value 130 is two standard deviations above the mean.

### 11.3 Python example

```python
from scipy.stats import norm

mu = 100
sigma = 15

# Probability that X is less than or equal to 130
print("P(X <= 130):", norm.cdf(130, loc=mu, scale=sigma))

# Probability that X is greater than 130
print("P(X > 130):", norm.sf(130, loc=mu, scale=sigma))

# Standardize a value
z = (130 - mu) / sigma
print("Z-score:", z)
```

The first calculation gives a probability of approximately 0.9772. The second gives approximately 0.0228.

### 11.4 Normal distribution in Machine Learning

Normal distributions appear in probability models, statistical assumptions, noise models, and some approaches to parameter initialization.

However, not every feature or prediction error in Machine Learning follows a Normal distribution. The suitability of this distribution depends on the problem and the assumptions being made.

## 12. Law of Large Numbers

The Law of Large Numbers states that, under suitable conditions, the sample average tends to approach the expected value as the number of independent observations increases.

For example, a fair coin has a probability of 0.5 of landing heads.

If we toss it only four times, we might observe three heads. The observed proportion is:

\[
\frac34=0.75
\]

This does not contradict the probability of 0.5. With only a few tosses, random variation can be substantial.

As the number of independent tosses increases, the proportion of heads tends to get closer to 0.5.

For independent and identically distributed observations with a finite expected value, the sample mean converges to the population mean under the Law of Large Numbers.

### Python simulation

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Simulate 10,000 fair coin tosses
tosses = np.random.binomial(1, 0.5, size=10_000)

# Calculate the running proportion of heads
running_proportion = np.cumsum(tosses) / np.arange(1, len(tosses) + 1)

plt.plot(running_proportion)
plt.axhline(y=0.5, linestyle="--", label="Expected proportion")
plt.xlabel("Number of Tosses")
plt.ylabel("Proportion of Heads")
plt.title("Law of Large Numbers")
plt.legend()
plt.grid(True)
plt.show()
```

The graph should show the running proportion fluctuating, especially at the beginning, and generally stabilizing closer to 0.5 as more tosses are included.

The Law of Large Numbers does not say that the proportion must move closer to 0.5 after every additional toss. It describes long-run convergence, not a guarantee of continuous improvement.

### Why it matters

The Law of Large Numbers helps explain why averages and estimated probabilities become more reliable with larger samples under suitable assumptions.

This is relevant to simulations, empirical probability estimates, and evaluating model performance on sufficiently representative datasets.

## 13. Central Limit Theorem

The Central Limit Theorem (CLT) explains why the distribution of sample means often becomes approximately Normal as the sample size increases, even when the original population is not Normal, provided suitable assumptions hold.

Suppose observations are independent and identically distributed, with finite mean \(\mu\) and finite, nonzero variance \(\sigma^2\).

For a sample of size \(n\), the sample mean is:

\[
\bar{X}=\frac{1}{n}\sum_{i=1}^{n}X_i
\]

As \(n\) increases, the standardized sample mean:

\[
\boxed{
\frac{\bar{X}-\mu}{\sigma/\sqrt{n}}
}
\]

converges in distribution to a standard Normal random variable under the usual CLT assumptions.

For sufficiently large samples, we often use the approximation:

\[
\bar{X}\approx N\left(\mu,\frac{\sigma^2}{n}\right)
\]

The quality of the approximation depends on the underlying distribution and the sample size. There is no single sample size that guarantees a good approximation for every distribution.

### 13.1 Standard error

The standard deviation of the sampling distribution of the sample mean is called the standard error.

For independent observations with population standard deviation \(\sigma\):

\[
\boxed{SE(\bar{X})=\frac{\sigma}{\sqrt{n}}}
\]

As the sample size increases, the standard error decreases.

This describes the variability of the sample mean across repeated samples, not the spread of individual observations.

### 13.2 Python simulation

The following example uses a skewed distribution to show how sample means can become approximately Normal as the sample size increases.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate a large population from an exponential distribution
population = np.random.exponential(scale=2, size=100_000)

sample_sizes = [5, 30, 100]
number_of_samples = 5000

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, n in zip(axes, sample_sizes):
    sample_means = []

    for _ in range(number_of_samples):
        sample = np.random.choice(population, size=n, replace=True)
        sample_means.append(np.mean(sample))

    ax.hist(sample_means, bins=40, density=True)
    ax.set_title(f"Sample Size = {n}")
    ax.set_xlabel("Sample Mean")
    ax.set_ylabel("Density")

plt.tight_layout()
plt.show()
```

The exponential population is right-skewed, but the distribution of its sample means generally becomes more symmetric as the sample size increases.

The CLT is important in statistical inference, confidence intervals, and understanding the behaviour of averages computed from data.

### Law of Large Numbers versus Central Limit Theorem

These concepts are related but different.

- **Law of Large Numbers:** Explains why the sample mean approaches the population mean as the sample size increases.
- **Central Limit Theorem:** Explains the approximate shape of the sampling distribution of the sample mean after suitable standardization.

One concerns convergence toward the expected value; the other concerns the distribution of the sample mean.

## 14. Simulating Probability with NumPy

Probability can be studied analytically using formulas or empirically by running simulations.

A simulation allows us to repeat a random experiment many times and estimate probabilities from the observed results.

For an event \(A\), the estimated probability is:

\[
\boxed{
\hat{P}(A)=\frac{\text{Number of times }A\text{ occurs}}
{\text{Total number of trials}}
}
\]

As the number of independent trials increases, this estimate often approaches the true probability under suitable conditions.

### 14.1 Simulating coin tosses

```python
import numpy as np

np.random.seed(42)

# Simulate 1,000 fair coin tosses
tosses = np.random.binomial(1, 0.5, size=1000)

heads = np.sum(tosses)
tails = len(tosses) - heads

print("Heads:", heads)
print("Tails:", tails)
print("Estimated probability of heads:", heads / len(tosses))
```

The estimated probability may not be exactly 0.5 because the experiment is random.

The seed makes the sequence reproducible for the same random-number generator and environment.

### 14.2 Simulating a die roll

```python
import numpy as np

np.random.seed(42)

# Simulate 10,000 rolls of a fair six-sided die
rolls = np.random.randint(1, 7, size=10_000)

# Estimate the probability of rolling a six
estimated_probability = np.mean(rolls == 6)

print("Estimated P(X = 6):", estimated_probability)
print("Theoretical P(X = 6):", 1 / 6)
```

The estimated probability should generally be near \(1/6\), though it will usually differ slightly due to random variation.

### 14.3 Visualizing a Binomial distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.5

values = np.arange(n + 1)
probabilities = binom.pmf(values, n, p)

plt.bar(values, probabilities)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.xticks(values)
plt.grid(axis="y")
plt.show()
```

This plot shows the probability of obtaining each possible number of successes in ten independent trials, where each trial has a success probability of 0.5.

### Why simulations are useful

Simulations help us:

- Check whether theoretical calculations are reasonable.
- Explore random behaviour.
- Understand probability distributions.
- Observe how sample size affects estimates.
- Build intuition before applying probability to real datasets.

A simulation provides an estimate, not a replacement for a correct theoretical calculation when an exact answer is available.

## 15. Applications of Probability in Machine Learning

Probability is used in several areas of Machine Learning.

### 15.1 Classification

A classification model may estimate the probability that an observation belongs to each class.

For example:

\[
P(\text{Cat}\mid\text{Image})=0.85
\]

\[
P(\text{Dog}\mid\text{Image})=0.15
\]

These values represent the model's estimated class probabilities. The model might select the cat class because it has the higher probability, depending on the classification procedure.

### 15.2 Naive Bayes

Naive Bayes classifiers apply Bayes' theorem to estimate class probabilities from observed features.

They are often used in spam detection, document classification, and text analysis.

The "naive" assumption is that features are conditionally independent given the class. This assumption is often imperfect, but the method can still be useful in practice.

### 15.3 Uncertainty estimation

Probability distributions help describe uncertainty in predictions and model outputs.

For example, a probabilistic model may provide a distribution over possible outcomes rather than only a single point prediction.

The quality of the uncertainty estimates depends on the model, data, and assumptions.

### 15.4 Model evaluation

Probability is also relevant to evaluating classifiers. Metrics such as log loss assess predicted probabilities, while other metrics assess the final predicted classes.

For a binary outcome \(y\in\{0,1\}\) and predicted probability \(p\), the binary cross-entropy loss for one observation is:

\[
\boxed{
L=-[y\log(p)+(1-y)\log(1-p)]
}
\]

This loss penalizes predictions that assign low probability to the observed outcome. In practice, probabilities are handled carefully near 0 and 1 to avoid numerical issues.

### 15.5 Data modelling and simulation

Probability distributions can be used to model random variables, generate synthetic observations, and study possible outcomes.

For example, a model may use a Normal distribution to represent measurement noise when that assumption is reasonable.

Probability helps make assumptions explicit so that we can understand when a model is appropriate and where its limitations may lie.

## 16. Summary of Important Formulas

| Concept | Formula |
|---|---|
| Equally likely outcomes | \(P(A)=\frac{\text{Favourable outcomes}}{\text{Total outcomes}}\) |
| Complement | \(P(A^c)=1-P(A)\) |
| Addition rule | \(P(A\cup B)=P(A)+P(B)-P(A\cap B)\) |
| Conditional probability | \(P(A\mid B)=\frac{P(A\cap B)}{P(B)}\) |
| Multiplication rule | \(P(A\cap B)=P(A)P(B\mid A)\) |
| Independent events | \(P(A\cap B)=P(A)P(B)\) |
| Bayes' theorem | \(P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}\) |
| Expected value, discrete | \(E[X]=\sum_x xP(X=x)\) |
| Variance | \(\operatorname{Var}(X)=E[X^2]-(E[X])^2\) |
| Standard deviation | \(\sigma=\sqrt{\operatorname{Var}(X)}\) |
| Bernoulli mean | \(E[X]=p\) |
| Bernoulli variance | \(\operatorname{Var}(X)=p(1-p)\) |
| Binomial PMF | \(P(X=k)=\binom nk p^k(1-p)^{n-k}\) |
| Binomial mean | \(E[X]=np\) |
| Binomial variance | \(\operatorname{Var}(X)=np(1-p)\) |
| Uniform mean | \(E[X]=\frac{a+b}{2}\) |
| Uniform variance | \(\operatorname{Var}(X)=\frac{(b-a)^2}{12}\) |
| Standardization | \(Z=\frac{X-\mu}{\sigma}\) |
| Standard error of a sample mean | \(SE(\bar X)=\frac{\sigma}{\sqrt n}\) |
